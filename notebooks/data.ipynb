```python
import pandas as pd

from nba_api.stats.endpoints import (
    leaguegamefinder,
    leaguedashplayerstats,
    leaguedashteamstats,
    scoreboardv2,
)

pd.set_option("display.max_columns", 100)

SEASON = "2025-26"
SEASON_TYPE = "Regular Season"
PLAYOFF_SEASON_TYPE = "Playoffs"
```

## Team game logs

```python
team_game_logs_df = leaguegamefinder.LeagueGameFinder(
    player_or_team_abbreviation="T",
    season_nullable=SEASON,
    season_type_nullable=SEASON_TYPE,
).get_data_frames()[0]

team_game_logs_df.head()
```

## Playoff game logs

```python
playoff_game_logs_df = leaguegamefinder.LeagueGameFinder(
    player_or_team_abbreviation="T",
    season_nullable=SEASON,
    season_type_nullable=PLAYOFF_SEASON_TYPE,
).get_data_frames()[0]

playoff_game_logs_df.head()
```

## Team advanced stats

```python
team_advanced_df = leaguedashteamstats.LeagueDashTeamStats(
    season=SEASON,
    season_type_all_star=SEASON_TYPE,
    measure_type_detailed_defense="Advanced",
    per_mode_detailed="PerGame",
).get_data_frames()[0]

team_advanced_df.head()
```

## Team base stats

```python
team_base_df = leaguedashteamstats.LeagueDashTeamStats(
    season=SEASON,
    season_type_all_star=SEASON_TYPE,
    measure_type_detailed_defense="Base",
    per_mode_detailed="PerGame",
).get_data_frames()[0]

team_base_df.head()
```

## Player aggregate stats

```python
player_base_df = leaguedashplayerstats.LeagueDashPlayerStats(
    season=SEASON,
    season_type_all_star=SEASON_TYPE,
    measure_type_detailed_defense="Base",
    per_mode_detailed="PerGame",
).get_data_frames()[0]

player_base_df.sort_values(by="GP", ascending=False).head()
```

## Daily schedule and standings context

Scoreboard data gives home/away matchups, game status, line scores, series standings during playoffs, and conference standings for a date.

```python
GAME_DATE = "2026-04-14"

scoreboard = scoreboardv2.ScoreboardV2(game_date=GAME_DATE)
scoreboard_frames = scoreboard.get_data_frames()

game_header_df = scoreboard_frames[0]
line_score_df = scoreboard_frames[1]
series_standings_df = scoreboard_frames[6]
east_standings_df = scoreboard_frames[4]
west_standings_df = scoreboard_frames[5]

game_header_df.head()
```
